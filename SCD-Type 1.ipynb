```python
df_clean = spark.table("dev.silver.silver_customers")
```

```python
df_clean.createOrReplaceTempView("customer_staging")
```

```python
%sql
CREATE OR REPLACE TABLE dev.silver.final_customers (
    customer_id STRING,
    first_name STRING,
    last_name STRING,
    date_of_birth DATE,
    email STRING,
    phone BIGINT,
    address STRING,
    city STRING,
    state STRING,
    postal_code INT,
    customer_segment STRING,
    customer_status STRING,
    registration_date DATE,
    updated_at DATE,
    file_name STRING,
    file_path STRING
)
USING DELTA;
```

```python
%sql
MERGE INTO dev.silver.final_customers AS target
USING customer_staging AS source
ON target.customer_id = source.customer_id
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *
```
