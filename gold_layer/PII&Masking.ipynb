```python
%sql
-- Original Data Sample
SELECT 
    'Original Data' AS data_type,
    customer_id,
    first_name,
    last_name,
    email AS original_email,
    phone AS original_phone
FROM dev.silver.final_customers
LIMIT 5;
```

```python
%sql
CREATE OR REPLACE TABLE dev.silver.final_customers_masked AS
SELECT 
    customer_id,
    first_name,
    last_name,
    date_of_birth,
    
    -- Email Masking: Mask username part, show domain
    -- Example: john.doe@example.com → ****@example.com
    CONCAT('****', SUBSTRING(email, INSTR(email, '@'), LENGTH(email))) AS email,
    
    -- Phone Masking: Show only last 4 digits, mask rest with asterisks
    -- Example: 9876543210 → ******3210
    CONCAT(
        LPAD('', LENGTH(CAST(phone AS STRING)) - 4, '*'),
        RIGHT(CAST(phone AS STRING), 4)
    ) AS phone,
    address
FROM dev.silver.final_customers;
```
